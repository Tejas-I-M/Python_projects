```python
import matplotlib.pyplot as plt

x= np.arange(1,11)
y=np.arange(10,110,10)
plt.figure(figsize=(5,3))
plt.plot(x,y,"r--")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
theta=np.pi
x=np.arange(0,2*theta,0.1)
y=np.sin(x)

plt.figure(figsize=(5,3))
plt.plot(x,y,"r--")
plt.show()
```
