```python
import matplotlib.pyplot as plt
plt.style.use("dark_background")
```

```python
theta=np.pi
```

```python
np.sin(theta/2)
```

```python
np.tan(0)
```

```python
x=np.arange(1,11)
y=np.arange(10,110,10)
plt.figure(figsize=(5,3))
plt.plot(x,y,'r--')
plt.title("Line Plot")
plt.show()
```

```python
x_sin = np.arange(0,2*theta,0.001)
y_sin = np.sin(x_sin)

plt.figure(figsize=(5,3))
plt.plot(x_sin,y_sin,'g-')
plt.title("Sine Wave")
plt.show()
```
