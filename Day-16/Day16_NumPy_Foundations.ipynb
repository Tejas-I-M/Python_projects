```python
student_marks = np.array([
    [78, 85, 90, 67],
    [45, 56, 40, 50],
    [88, 92, 95, 91],
    [60, 65, 70, 72],
    [30, 35, 28, 40]
])

print("Dataset:\n")
print(student_marks)
```

```python
student_avg = np.mean(student_marks, axis=1)

print("\nStudent Average:")
print(student_avg)
```

```python
subject_avg = np.mean(student_marks, axis=0)

print("\nSubject Average:")
print(subject_avg)
```

```python
total_marks = np.sum(student_marks, axis=1)
print(total_marks)
topper_index = np.argmax(total_marks)

print("\nTopper Index:", topper_index)
```

```python
highest_marks = np.max(student_marks, axis=0)
print("\nHighest Marks in Each Subject:"   )
print(highest_marks)
```

```python
min_val = np.min(student_marks)
max_val = np.max(student_marks)
print("\nMinimum Value:", min_val)
print("Maximum Value:", max_val)
print(student_marks)
normalized = (student_marks - min_val) / (max_val - min_val)

print(normalized)
```
